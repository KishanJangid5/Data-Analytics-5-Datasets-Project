## Missing Value Handling

After removing duplicate product records, the dataset contains missing values in
some optional product information fields such as discount, customer comments,
seller information, videos, and product variations.

These missing values were retained rather than removed or replaced with
artificial values because the absence of information does not necessarily mean
that the product has no such feature. Keeping the values as missing preserves
the original information and avoids introducing assumptions into the analysis.

### Discount and Price Analysis

Correlation analysis showed a strong positive relationship between initial price and final price (0.80), meaning higher-priced products generally tend to have higher final prices.

The correlation between discount and final price was very weak (-0.07), indicating that discount percentage has almost no linear relationship with final price in this dataset.

The correlation between discount and initial price was also weak (0.14).

An important data-quality observation is that some products have a discount value even though their initial price and final price are identical. Therefore, the price fields should be interpreted carefully and should not be assumed to perfectly represent the actual discounted selling price.

# Final Summary — Ecommerce Dataset

## Dataset Overview

The Ecommerce Dataset contains product information collected across 98 category-specific CSV files.

All 98 files had the same 24-column structure. After combining the files, the dataset contained 2,000 rows.

During data cleaning, 1,000 exact duplicate rows were identified and removed, resulting in 1,000 unique products.

## Data Cleaning

The following cleaning steps were performed:

- Combined all 98 CSV files into one DataFrame.
- Verified that all files had the same column structure.
- Removed exact duplicate records.
- Converted `final_price` from text to numeric format.
- Checked missing values across all columns.
- Kept missing values in optional fields instead of replacing them with artificial values.
- Reviewed the `discount` column and retained its missing values because a missing discount does not necessarily mean a 0% discount.

## Category Analysis

The dataset contains 97 product categories.

The categories with the highest number of products were:

- Tops — 122 products
- Dresses — 100 products
- Shirts — 97 products
- Jeans — 57 products
- Sports-shoes — 51 products

The top three categories alone represented 31.9% of the 1,000 unique products.

## Price Analysis

The average final price across products varied considerably between categories.

Among categories with at least 10 products:

- Watches had the highest average final price at approximately ₹5,393.
- Sports-shoes followed with approximately ₹4,228.
- Sweatshirts averaged approximately ₹2,839.
- Jeans averaged approximately ₹2,174.
- Kurtas averaged approximately ₹1,919.

Categories with very few products were excluded from the reliable category comparison because their averages could be misleading.

## Rating Analysis

The overall average product rating was approximately 3.62 out of 5.

Among categories with at least 10 products:

- Watches had the highest average rating at approximately 4.15.
- Sweaters followed at approximately 4.15.
- Sports-shoes averaged approximately 4.12.
- Handbags averaged approximately 4.10.

This shows that some categories combine relatively high prices with strong customer ratings.

## Discount and Price Analysis

Correlation analysis showed:

- Initial price vs final price: approximately 0.80
- Discount vs final price: approximately -0.07
- Discount vs initial price: approximately 0.14

The strong positive relationship between initial price and final price indicates that higher-priced products generally have higher final prices.

The very weak relationship between discount and final price indicates that discount percentage alone does not strongly determine the final price.

Some products also contained discount values while their initial and final prices were identical. This was identified as a potential data-quality issue.

## Rating and Review Analysis

The correlation between rating and ratings count was approximately 0.13.

This represents a very weak positive relationship.

Therefore, products with more reviews do not necessarily have much higher ratings.

## Product-Level Analysis

Products with high ratings were analyzed using minimum review-count thresholds.

When requiring at least 10 reviews, several products had ratings between 4.8 and 5.0.

A stricter analysis using at least 50 reviews showed that:

- Arrow shirts had a rating of 4.7 with 64 reviews.
- Van Heusen briefs had a rating of 4.6 with 192 reviews.
- SASSAFRAS dresses had a rating of 4.5 with 4,441 reviews.

This demonstrated that rating count should also be considered when evaluating highly rated products.

## Customer Engagement by Category

Dresses had the highest total number of reviews with 15,462 reviews.

Other highly reviewed categories included:

- Shirts — 6,085 reviews
- Tops — 5,077 reviews
- Jeans — 4,837 reviews
- Tshirts — 4,715 reviews

This indicates that dresses generated substantially more customer review activity than other categories in this dataset.

## Key Business Insights

1. Dresses have the highest number of products and the highest total number of reviews.
2. Watches have one of the highest average ratings and the highest average price among sufficiently large categories.
3. Higher initial prices generally correspond to higher final prices.
4. Discount percentage has very little relationship with final price.
5. Rating and review count have only a weak relationship.
6. A high rating should be considered together with review count rather than alone.
7. Some price and discount fields may contain inconsistencies and should be treated carefully.

## Conclusion

The Ecommerce dataset was cleaned, explored, and analyzed using Python and Pandas.

The analysis covered product categories, pricing, discounts, ratings, review counts, customer engagement, and data-quality issues.

The cleaned dataset provides a useful foundation for further analysis using SQL and for creating a business dashboard in Power BI.
